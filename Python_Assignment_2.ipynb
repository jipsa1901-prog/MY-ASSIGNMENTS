# Python Assignment 2: Data Structures & Conditional Statements

## 1. List Creation

```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Arun", "Bala", "Chitra", "Deepa", "Eswar"]
print(age_list)
print(name_list)
```

## 2. List Operations

```python
name_list.append("Yazhini")
print("After append:", name_list)

age_list.insert(2, 30)
print("After insert:", age_list)

name_list.remove("Yazhini")
print("After remove:", name_list)

popped = age_list.pop()
print("Popped:", popped, "->", age_list)

age_list.extend([29, 30, 26])
print("After extend:", age_list)

age_list.sort(reverse=True)
print("Descending:", age_list)

print("Max:", max(age_list))
print("Min:", min(age_list))
print("Sum:", sum(age_list))
```

## 3. Accessing List Elements

```python
print("First:", name_list[0])
print("Last:", name_list[-1])
print("Index 2 to 4:", name_list[2:5])
print("Reverse:", name_list[::-1])
```

## Dictionary

```python
student_marks = {"Arun": 78, "Bala": 85, "Chitra": 92, "Deepa": 67, "Eswar": 74}
print(student_marks)

print("Chitra's mark:", student_marks["Chitra"])

student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

student_marks["Deepa"] = 82
print("After updating Deepa:", student_marks)

print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

## Sets

```python
my_set = set(['a','e','i','o','u','a','a','i'])
print(my_set)
```

**Explanation:** A set stores only unique elements, so the duplicate 'a' and 'i' are removed. Sets are also unordered, so the print order may differ.

```python
try:
    my_set[4] = 's'
except TypeError as e:
    print('Error:', e)
```

**Explanation:** Sets are unordered and do not support indexing, so item assignment with `[]` raises a `TypeError`.

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print('Union:', set1 | set2)
print('Intersection:', set1 & set2)
```

## Performance Category Program

```python
score = float(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score. Please enter a value between 0 and 10.")
elif score > 7:
    print("Above Average: Excellent work! Keep up the great performance.")
elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")
else:
    print("Below Average: Need to improve your performance, consistent practice will lead to better results.")
```
