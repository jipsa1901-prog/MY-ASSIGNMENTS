# Python Assignment 1: Data Structures - Strings & Tuples

## 1. String Concatenation

```python
string1 = "Hello "
name = input("Enter your Name: ")
greeting = string1 + name
print(greeting)
```

```python
string3 = ", welcome to Python programming"
full_string = greeting + string3
print(full_string)
```

## 2. String Slicing and Indexing

```python
s = full_string

# a. First character
print("a.", s[0])

# b. Last character
print("b.", s[-1])

# c. First 5 characters
print("c.", s[:5])

# d. Last 11 characters
print("d.", s[-11:])

# e. String in reverse
print("e.", s[::-1])

# f. Word "Python" using slicing
start = s.find("Python")
print("f.", s[start:start + len("Python")])
```

## 3. String Methods

```python
strM = "Python beginner tutorial"

# a. Uppercase
print("a.", strM.upper())

# b. Lowercase
print("b.", strM.lower())

# c. Capitalize -> back to the original form
print("c.", strM.lower().capitalize())

# d. Count of character 't'
print("d.", strM.count("t"))

# e. Replace "Python" with "Machine Learning"
print("e.", strM.replace("Python", "Machine Learning"))
```

## Tuples (Creation, Modification and Access)

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)

# a. Concatenate
t_combine = t1 + t2
print("a.", t_combine)

# b. Repeat 3 times
print("b.", t_combine * 3)

# c. 3rd element
print("c.", t_combine[2])

# d. First three elements
print("d.", t_combine[:3])

# e. Last three elements
print("e.", t_combine[-3:])
```
